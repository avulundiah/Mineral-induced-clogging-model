```python
from pathlib import Path
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import brentq

OUTPUT = Path(__file__).resolve().parent
DELTA, GAMMA, KAPPA, EXPONENT, X0 = 1.0, 1.0, 1.0, 5.0, 0.10


def positive_loading(rzero):
    if rzero <= 1:
        return np.nan
    lam = 1.0 / ((1.0 + KAPPA) * rzero)
    substrate = lam * KAPPA / (1.0 - lam)
    h = DELTA * (1.0 - substrate) / (GAMMA * lam)
    return brentq(
        lambda x: x - h * (1.0 + X0 - x) ** EXPONENT,
        0.0, 1.0 + X0, xtol=1e-13,
    )


def main():
    rzero = np.linspace(1.001, 4.0, 600)
    loading = np.array([positive_loading(r) for r in rzero])
    fig, ax = plt.subplots(figsize=(6.7, 3.6))
    ax.plot([0.2, 1.0], [0, 0], color="#145c9e", lw=2.4,
            label="Mineral-free: stable")
    ax.plot([1.0, 4.0], [0, 0], color="#555555", lw=1.8, ls="--",
            label="Mineral-free: unstable")
    ax.plot(rzero, loading, color="#a12c40", lw=2.4,
            label="Positive equilibrium: stable")
    ax.axvline(1.0, ls=":", color="#777777", lw=1)
    ax.scatter([1.0], [0], color="black", s=23, zorder=5)
    ax.set_xlim(.2, 4.0)
    ax.set_ylim(-.015, .57)
    ax.set_xlabel(r"Persistence number $\mathcal{R}_0$")
    ax.set_ylabel(r"Equilibrium mineral loading $x^*$")
    ax.legend(loc="lower right", frameon=False, fontsize=9)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.grid(alpha=.18, lw=.65)
    fig.tight_layout()
    fig.savefig(OUTPUT / "equilibrium_bifurcation_revised.pdf")
    fig.savefig(OUTPUT / "equilibrium_bifurcation_revised.png", dpi=450)
    plt.close(fig)
    print(f"At R0=2, x*={positive_loading(2):.8f}")


if __name__ == "__main__":
    main()
```
