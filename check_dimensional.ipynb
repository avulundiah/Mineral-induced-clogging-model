```python
import numpy as np
from scipy.integrate import solve_ivp
from scipy.optimize import minimize_scalar,brentq
r=.16;gamma=540;delta=r*gamma;kappa=.3;lam=.35;n=5;x0=.2;T=30
for fb in (True,False):
 def fun(t,y):
  s,x=y;f=s/(kappa+s);return [delta*((1+x0-x)**n if fb else 1)*(1-s)-gamma*f*x,(f-lam)*x]
 z=solve_ivp(fun,(0,T),(.9,x0),method='LSODA',dense_output=True,rtol=1e-10,atol=1e-12)
 p=lambda t:(1+x0-z.sol(t)[1])**n
 tg=np.linspace(0,T,1001);j=int(np.argmin(p(tg)));lo=tg[max(j-1,0)];hi=tg[min(j+1,1000)];fit=minimize_scalar(p,bounds=(lo,hi),method='bounded')
 minimum=min(float(fit.fun), float(p(30)), float(p(0)))
 edges=[]
 for a,b in zip(tg[:-1],tg[1:]):
  if (p(a)-.4)*(p(b)-.4)<0:edges.append(brentq(lambda x:p(x)-.4,a,b))
 cuts=[0]+edges+[T];dur=sum(b-a for a,b in zip(cuts[:-1],cuts[1:]) if p(.5*(a+b))<.4)
 print('fb',fb,'min',minimum,'argmin',fit.x,'k30',p(30),'duration',dur,'crossings',edges,'pressure_ratio_at_min',1/minimum,'x30',z.sol(30)[1])
 # eigenvalues at eq
```
