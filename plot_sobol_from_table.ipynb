```python
from pathlib import Path
import csv,numpy as np
import matplotlib;matplotlib.use('Agg')
import matplotlib.pyplot as plt
ROOT=Path(__file__).resolve().parent
with (ROOT/'global_sensitivity_indices_reduced.csv').open() as f: rows=list(csv.DictReader(f))
keys=['r','kappa','lambda','n']; rows=sorted(rows,key=lambda x:keys.index(x['Parameter']))
first=np.array([float(x['FirstOrder']) for x in rows]);total=np.array([float(x['TotalOrder']) for x in rows]);
errf=np.array([(float(x['FirstCI_U'])-float(x['FirstCI_L']))/2 for x in rows]);errt=np.array([(float(x['TotalCI_U'])-float(x['TotalCI_L']))/2 for x in rows]);
assert np.all(first<=total+1e-12),'Analytical Sobol ordering violated'
fig,(ax,bx)=plt.subplots(1,2,figsize=(8.3,3.45),gridspec_kw={'width_ratios':[3.4,1]})
labels=[r'$r=\delta/\gamma$',r'$\kappa$',r'$\lambda$',r'$n$']
x=np.arange(4);w=.34
ax.bar(x-w/2,first,w,yerr=errf,capsize=3,label='First order')
ax.bar(x+w/2,total,w,yerr=errt,capsize=3,label='Total order')
ax.set_xticks(x,labels);ax.set_ylim(0,.72);ax.set_ylabel('Sobol index')
ax.legend(frameon=False,fontsize=8);ax.grid(axis='y',alpha=.2)
bx.bar([0,1],[first[1],total[1]],width=.56,yerr=[errf[1],errt[1]],capsize=4)
bx.set_ylim(.005,.009);bx.set_xticks([0,1],['First','Total']);bx.set_title(r'$\kappa$ (expanded)',fontsize=9)
bx.grid(axis='y',alpha=.2)
fig.tight_layout();fig.savefig(ROOT/'global_sensitivity_reduced.pdf');fig.savefig(ROOT/'global_sensitivity_reduced.png',dpi=400);plt.close(fig)
```
