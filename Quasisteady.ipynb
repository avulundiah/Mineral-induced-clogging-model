```python
from pathlib import Path
import numpy as np
from scipy.integrate import solve_ivp
from scipy.optimize import brentq
import matplotlib;matplotlib.use('Agg')
import matplotlib.pyplot as plt
OUT=Path(__file__).resolve().parent
r=.16;gamma=540.;delta=gamma*r; kappa=.3;lam=.35;n=5.;x0=.2;c=1+x0;T=30.

def sq(x):
 a=r*max(c-x,0)**n
 if a==0:return 0.
 b=x-a*(1-kappa)
 return (2*a*kappa)/(b+np.sqrt(b*b+4*a*a*kappa)) if b>0 else (-b+np.sqrt(b*b+4*a*a*kappa))/(2*a)

def full(t,y):
 s,x=y;f=s/(kappa+s)
 return [delta*max(c-x,0)**n*(1-s)-gamma*f*x,(f-lam)*x]

def reduced(t,x):
 s=sq(float(x[0]));return [(s/(kappa+s)-lam)*x[0]]
sol=solve_ivp(full,(0,T),(.9,x0),method='LSODA',dense_output=True,rtol=1e-10,atol=1e-12)
qsol=solve_ivp(reduced,(0,T),(x0,),method='DOP853',dense_output=True,rtol=1e-11,atol=1e-13)
t=np.linspace(0,T,1201);kf=(c-sol.sol(t)[1])**n;kq=(c-qsol.sol(t)[0])**n
print('max K discrepancy',np.max(abs(kf-kq)),'K30 full',kf[-1],'QSS',kq[-1]);print('s_q initial',sq(x0),'s0',.9)
fig,ax=plt.subplots(figsize=(6.9,3.8))
ax.plot(t,kf,label='Full two-state model',lw=2)
ax.plot(t,kq,'--',label='Fast-substrate scalar reduction',lw=1.6)
ax.set(xlabel='Physical time (days)',ylabel=r'Normalized permeability $k/k_0$')
ax.legend(frameon=False,fontsize=9);ax.grid(alpha=.25)
fig.tight_layout();fig.savefig(OUT/'quasisteady_comparison.pdf');plt.close(fig)
```
