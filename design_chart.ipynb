```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

KAPPA = 1.0                              
X0 = 0.0                                 
EPS_A = 0.10                             
N_LIST = [4, 5, 6, 8, 9.7, 12]           
N_B = 5                                  
EPS_LIST = [0.5, 0.2, 0.1, 0.05, 0.01]   
OUT = "/mnt/user-data/outputs"

LAM_C = 1.0 / (KAPPA + 1.0)              


def a_crit(eps, n, x0=X0):
    return (1.0 + x0 - eps ** (1.0 / n)) / eps


def boundary(lam, eps, n, kappa=KAPPA, x0=X0):
    return a_crit(eps, n, x0) * lam * (1.0 - lam) / (1.0 - (kappa + 1.0) * lam)
plt.rcParams.update({
    "font.family": "serif",
    "mathtext.fontset": "cm",
    "font.size": 9,
    "axes.labelsize": 10,
    "legend.fontsize": 8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "axes.linewidth": 0.8,
})

lam = np.linspace(0.004, LAM_C * 0.9995, 2000)
colors = ["#0072B2", "#009E73", "#E69F00", "#D55E00", "#CC79A7", "#444444"]
styles = ["-", "--", "-.", ":", (0, (5, 1, 1, 1, 1, 1)), (0, (1, 1))]
ymin, ymax = 1e-2, 1e2

fig, axes = plt.subplots(1, 2, figsize=(7.2, 3.5), sharey=True)


def decorate(ax):
    ax.set_yscale("log")
    ax.set_xlim(0, LAM_C * 1.04)
    ax.set_ylim(ymin, ymax)
    ax.axvspan(LAM_C, LAM_C * 1.04, color="0.88", lw=0)
    ax.axvline(LAM_C, color="k", lw=0.9, ls=(0, (2, 2)))
    ax.set_xlabel(r"mineral removal rate $\lambda=k_d/\mu_{\max}$")
    ax.grid(True, which="major", color="0.9", lw=0.5)
    ax.tick_params(direction="in", which="both", top=False)
    # secondary axis in R0
    ax2 = ax.twiny()
    ax2.set_xlim(ax.get_xlim())
    R_ticks = np.array([1.25, 1.5, 2, 3, 5, 10])
    ax2.set_xticks(1.0 / ((KAPPA + 1.0) * R_ticks))
    ax2.set_xticklabels([f"{r:g}" for r in R_ticks])
    ax2.tick_params(direction="in", labelsize=8)
    ax2.set_xlabel(r"invasion number $\mathcal{R}_0$", fontsize=9)


# ---- panel (a): several n, fixed eps
ax = axes[0]
for n, col, ls in zip(N_LIST, colors, styles):
    ax.plot(lam, boundary(lam, EPS_A, n), color=col, ls=ls, lw=1.6, label=rf"$n={n:g}$")
# shade the clogging region for the middle exponent n = 5
ax.fill_between(lam, boundary(lam, EPS_A, 5), ymax, color="#009E73", alpha=0.10, lw=0)
decorate(ax)
ax.set_ylabel(r"supply-to-capacity ratio $\delta/\gamma$")
ax.text(0.05, 55, r"clogged: $k^*/k_0\leq\varepsilon$", fontsize=8.5, color="#006B4E")
ax.text(0.11, 0.022, r"persistent, open: $k^*/k_0>\varepsilon$", fontsize=8.5, color="0.25")
ax.text(LAM_C * 1.02, 0.5, r"$\mathcal{R}_0\leq 1$: washout", rotation=90, fontsize=7.5,
        ha="center", va="center", color="0.35")
ax.set_title(rf"(a) $\varepsilon={EPS_A:g}$, varying $n$", fontsize=9, loc="left", pad=28)
ax.legend(loc="upper left", ncol=2, frameon=True, framealpha=0.95, edgecolor="0.8",
          bbox_to_anchor=(0.02, 0.89), columnspacing=1.0, handlelength=2.4)

# ---- panel (b): fixed n, several eps
ax = axes[1]
for eps, col, ls in zip(EPS_LIST, colors, styles):
    ax.plot(lam, boundary(lam, eps, N_B), color=col, ls=ls, lw=1.6, label=rf"$\varepsilon={eps:g}$")
decorate(ax)
ax.text(0.05, 55, r"clogged: $k^*/k_0\leq\varepsilon$", fontsize=8.5, color="#006B4E")
ax.text(LAM_C * 1.02, 0.5, r"$\mathcal{R}_0\leq 1$: washout", rotation=90, fontsize=7.5,
        ha="center", va="center", color="0.35")
ax.set_title(rf"(b) $n={N_B:g}$, varying $\varepsilon$", fontsize=9, loc="left", pad=28)
for eps, col in zip(EPS_LIST, colors):
    lam_lab = 0.07
    ax.text(lam_lab, boundary(lam_lab, eps, N_B) * 1.25, rf"$\varepsilon={eps:g}$", color=col,
            fontsize=8, ha="left", va="bottom",
            bbox=dict(boxstyle="round,pad=0.12", fc="white", ec="none", alpha=0.85))

fig.tight_layout()
fig.savefig(f"{OUT}/clogging_design_chart.pdf", bbox_inches="tight")
print("saved; A_c values (eps=%.2f, x0=%.2f):" % (EPS_A, X0),
      {n: round(a_crit(EPS_A, n), 3) for n in N_LIST})
```
